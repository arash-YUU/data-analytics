## Looping Through Arrays

```python
arrA = np.array([(11, 12, 13), (14, 15, 16)])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop prints each whole row multiplied by 2. The second loop goes inside each row and prints one doubled number at a time.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)
# x is the whole row, so the same doubled row prints once for each value in that row.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# ValueError: all input arrays must have the same shape.
# One array is 4 by 3 and the other is 2 by 6, so they cannot be stacked together.
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# ValueError: dimensions other than the joining axis must match.
# Joining columns needs the same number of rows, but these arrays have 4 and 2 rows.
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
```

```python
horizontal = np.hstack((arrA, arrA))  # Put the columns side by side
vertical = np.vstack((arrA, arrA))  # Put another set of rows underneath
depth = np.dstack((arrA, arrA))  # Stack the arrays along a third dimension
print(horizontal)
print(vertical)
print(depth)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# The result is a list of two 2-D arrays, each with 2 rows and 3 columns.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# The result is three 2-D arrays: one has 2 rows, and the other two have 1 row each.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")  # A remainder of zero means the number is even
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([(7, 6), (13, 5), (7, 3)])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
print(arrB)
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
