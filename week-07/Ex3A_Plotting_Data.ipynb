```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, marker="o")
plt.show()
```

```python
plt.plot(x, y, marker="o")
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

## Styling Plots

```python
plt.plot(x, y, linestyle="--", color="red")
plt.show()
```

```python
plt.plot(x, y, linestyle="--", color="red", marker="s", markersize=10,
         markerfacecolor="green", markeredgecolor="green")  # Keep the line red and change the square markers to green
plt.show()
```

```python
plt.plot(x, y, marker="^", color="purple")  # Triangle markers
plt.show()
```

```python
plt.plot(x, y, marker="D", color="blue")  # Diamond markers
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
plt.bar(products, sales)
plt.show()
```

```python
plt.bar(products, sales, color="skyblue", width=0.5)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y)
plt.show()
```

```python
s = np.random.rand(50) * 100
plt.scatter(x, y, s=s, c=y, cmap="viridis")  # y supplies the values used by the colormap
plt.show()
```

## Subplots

```python
fig, axes = plt.subplots(1, 2)  # One row with two separate plots
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
axes[0].plot(x, y)
```

```python
axes[1].hist(np.random.randn(1000))
```

```python
axes[0].set_title("Sine Wave")
axes[0].set_xlabel("X")
axes[0].set_ylabel("sin(x)")
axes[1].set_title("Random Numbers")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Frequency")
plt.show()
```
