## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.rand(4, 5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

shuffle changes the order of the rows in this 2-D array. The values inside each row stay together.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

I would use shuffle to change the existing array. I would use permutation when I want a shuffled copy and want to keep the original array unchanged.

## Mathematical Operations

```python
print(r_1)
print(r_2)
print(r_1.shape)
print(r_2.shape)
print(np.add(r_1, r_2))
print(np.add(r_1, r_2).shape)
```

```python
print(r_1)
print(r_3)
print(r_1.shape)
print(r_3.shape)
print(np.add(r_1, r_3))
print(np.add(r_1, r_3).shape)
```

```python
print(r_2)
print(r_4)
print(r_2.shape)
print(r_4.shape)
# ValueError: operands could not be broadcast together with shapes (1,2) (4,5).
# The last dimensions are 2 and 5, and neither is 1, so NumPy cannot line them up.
print(np.add(r_2, r_4))
```

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))
```

```python
# ValueError: operands could not be broadcast together with shapes (2,1) (4,5).
# The row counts 2 and 4 do not match, and neither of those dimensions is 1.
print(np.add(np.array([[1], [1]]), r_4))
```

Broadcasting compares dimensions from the right. Dimensions can match, or one can be 1 and repeat across the other. NumPy then adds matching positions, and the result uses the larger compatible size in each dimension.

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
# Both arrays use the order vegetables, fruits, snacks, beverages so matching categories add together.
print(bag1)
print(bag2)
```

```python
print(f"The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.")
```

```python
total_items = np.add(bag1, bag2)
print(f"There are {total_items[0]} vegetables, {total_items[1]} fruits, {total_items[2]} boxes of snacks, and {total_items[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
ben = np.array([0, 5, 5, 3, 4])  # Zero keeps Monday in the correct position
```

```python
carla = np.array([3, 6, 0, 3, 4])  # Zero keeps Wednesday in the correct position
```

```python
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
person_totals = np.sum(all_knocks, axis=1)
print(f"Ali knocked on {person_totals[0]} doors.")
print(f"Ben knocked on {person_totals[1]} doors.")
print(f"Carla knocked on {person_totals[2]} doors.")
print(f"The group knocked on {np.sum(all_knocks)} doors.")
```

```python
daily_totals = np.sum(all_knocks, axis=0)
rolling_votes = np.cumsum(daily_totals) * 0.5  # One estimated vote for every two door-knocks
print(rolling_votes)
```
