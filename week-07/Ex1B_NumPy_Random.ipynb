## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

```python
r_3 = np.random.rand(2, 1)
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

```python
r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(["Abdelouahab", "A'dariyon", "Cesar", "David", "Earl", "Ezekiel",
                     "Faiz", "Henry", "Joshua", "Matthew", "Meric", "Netsa"])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# This can choose a student more than once and leave others out, so it is not a reliable way to assign everyone to one group.
```

```python
call_on_weighted = np.random.choice(students, size=5,
    p=[0.30, 0, 0.15, 0.10, 0.10, 0.05, 0.05, 0.05, 0.05, 0.05, 0.05, 0.05])
# Abdelouahab has 30 percent, A'dariyon has zero, and Cesar has 15 percent.
# David and Earl each have 10 percent. Ezekiel, Faiz, Henry, Joshua, Matthew, Meric, and Netsa each have 5 percent.
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
