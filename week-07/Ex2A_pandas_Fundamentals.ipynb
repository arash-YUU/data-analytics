```python
import numpy as np
import pandas as pd
```

## Series

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
ser1 = pd.Series(aa)
print(ser1)
```

```python
ser2 = pd.Series(ab)
print(ser2)
```

```python
# TypeError: 'set' type is unordered. pandas does not accept a set directly as Series data.
ser3 = pd.Series(ac)
print(ser3)
```

The set cannot be used directly because it is unordered. The following correction converts it to a list so the later DataFrame steps can run.

```python
ser3 = pd.Series(list(ac))
print(ser3)
```

```python
ser4 = pd.Series(ad)
print(ser4)
```

```python
# TypeError: Index(...) must be called with a collection of some kind, 18 was passed.
# Separate arguments are treated as other Series parameters, not as four data values.
ser5 = pd.Series(17, 18, 19, 20)
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1)
df1
```

```python
df1.insert(1, "ser2", ser2)
df1.insert(2, "ser3", ser3)
df1
```

```python
df1.insert(3, "ser4", ser4)
# ser4 has word labels instead of row labels 0 to 3. pandas matches labels, so its values become NaN here.
df1
```

```python
df1.loc[4] = [5, 9, 12, 0]
df1
```
