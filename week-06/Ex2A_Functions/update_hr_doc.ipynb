## Lab 3. Employee data
This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [('0123', 'HR', 'Rebecca Yang', '69000'),
           ('0121', 'IT', 'Mark Blick', '67000'),
           ('0068', 'IT', 'Kahna Larsen', '112000'),
           ('0234', 'OPS', 'Jim Smith', '54000')]
```

```python
for i, employee in enumerate(hr_list):
    if employee[0] == "0121":  # Match the employee number so only Mark gets a new name
        hr_list[i] = (employee[0], employee[1], "Mark Blick-Hawley", employee[3])
        # Tuples cannot be changed in place, so the list gets a replacement tuple.
```

```python
print(hr_list)
```

```python
for i, employee in enumerate(hr_list):
    if employee[0] == "0234":  # Change only Jim's department and salary
        hr_list[i] = (employee[0], "CS", employee[2], "60000")
```

```python
print(hr_list)
```

```python
for employee_num, department, name, salary in hr_list:  # Take the four values from each employee tuple
    print(f"{employee_num} | {department} | {name} | {int(salary):,}")
    # Salary starts as text and must become a number for comma formatting.
```
