## Lab 1. Working with names - parsing strings

```python
test_name = 'John Doe'
```

```python
def parse_and_display(name):  # Find the first space and use it to separate the name
    for i in range(len(name)):
        if name[i] == " ":  # The space separates the first name from the rest
            first = name[:i]
            last = name[i + 1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
            break  # Stop after the first space so the same name is not printed again
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0])
```

```python
parse_and_display(test_list[1])
```

```python
parse_and_display(test_list[2])
```

```python
parse_and_display('Sayaeed Arashullah Akrami')
```

## Challenge: one, two, or three names

```python
def parse_and_display2(name):  # Handle a name with one, two, or three words
    parts = name.split()  # split handles extra spaces between words
    if not parts:
        print("No name was entered")
        return
    print(f"Full name: {name}")
    print(f"First name: {parts[0]}")
    if len(parts) == 1:  # A one-word name has no separate last name
        print("Last name: none")
    elif len(parts) == 2:  # With two words, the second word is the last name
        print(f"Last name: {parts[1]}")
    else:
        print(f"Middle name: {' '.join(parts[1:-1])}")
        print(f"Last name: {parts[-1]}")
```

```python
parse_and_display2('Lorde')
```

```python
parse_and_display2('Billie Eilish')
```

```python
parse_and_display2('Megan Thee Stallion')
```
