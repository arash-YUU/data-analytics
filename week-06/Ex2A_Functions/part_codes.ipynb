# Exercise 2.A Using built-in functions

## Lab 2. Parsing strings - part codes

```python
def get_supplier_code(part_code):  # Return the supplier part before the colon
    colon = part_code.index(":")
    return part_code[:colon]  # Everything before the colon is the supplier
```

```python
def get_product_num(part_code):  # Return the part number between the colon and dash
    colon = part_code.index(":")
    dash = part_code.index("-", colon)
    return part_code[colon + 1:dash]  # Leave out both separators
```

```python
def get_size(part_code):  # Return the size after the dash
    dash = part_code.index("-", part_code.index(":"))
    return part_code[dash + 1:]
```

```python
part_1 = 'ACME:123-L'
part_2 = 'DI:12345-M'
part_3 = 'ACE:1-12'
```

```python
print(f"For {part_1}, the supplier is {get_supplier_code(part_1)}. Part number {get_product_num(part_1)} is available in size {get_size(part_1)}.")
```

```python
print(f"For {part_2}, the supplier is {get_supplier_code(part_2)}. Part number {get_product_num(part_2)} is available in size {get_size(part_2)}.")
```

```python
print(f"For {part_3}, the supplier is {get_supplier_code(part_3)}. Part number {get_product_num(part_3)} is available in size {get_size(part_3)}.")
```

## Bonus: process a list of part codes

```python
part_codes = [part_1, part_2, part_3]
for part in part_codes:  # Pass each code through the same three functions
    print(f"For {part}, the supplier is {get_supplier_code(part)}. Part number {get_product_num(part)} is available in size {get_size(part)}.")
```
