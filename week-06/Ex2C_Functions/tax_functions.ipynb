## Lab 3. Tax functions
These functions use the rates supplied in the lab workbook.

```python
def get_soc_sec_tax(gross_pay):  # Calculate Social Security tax using the lab rate of 6.2 percent
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):  # Calculate Medicare tax using the lab rate of 1.45 percent
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):  # Choose the lab tax rate from the withholding code
    if withholding_code == 0:  # Code zero uses the highest rate in this table
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:  # Code three uses 18.5 percent; higher codes use the formula below
        rate = 0.185
    else:
        rate = 0.18 - (withholding_code - 4) * 0.005  # Each code above 4 lowers the rate by half a percent
    return gross_pay * rate
```

```python
gross_pay = 750
withholding_code = 0
print("Person 1")
print(f"Social Security tax: ${get_soc_sec_tax(gross_pay):.2f}")
print(f"Medicare tax: ${get_medicare_tax(gross_pay):.2f}")
print(f"Federal tax: ${get_federal_tax(gross_pay, withholding_code):.2f}")
```
