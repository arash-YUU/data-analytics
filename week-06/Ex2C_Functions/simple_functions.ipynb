# Exercise 2.C Creating and calling functions

## Lab 1. Defining custom functions
The calls below use example movie, music, and place choices.

```python
def favorite_things(name, movie, music):  # Return three lines using the values passed into the function
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician or band is {music}."
```

```python
print(favorite_things("Arash", "Interstellar", "Coldplay"))
```

```python
def why_im_here(purpose):  # Return the reason passed in, so the caller can print it
    return purpose
```

```python
print(why_im_here("I joined the Year Up United Data Analyst program to build my data skills and prepare for a career in analytics."))
```

```python
def favorite_place(place, why):  # Combine the place and reason into one sentence
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("the beach", "I enjoy the ocean view"))
```

```python
print(favorite_place("the park", "I can relax and walk outside"))
```
