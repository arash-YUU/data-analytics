## Lab 2. Working with datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
print(today.strftime("Today is %A, %B %d, %Y"))  # strftime turns the date into the requested text format
```

# My birthday

```python
birthday = datetime.date(2007, 6, 30)
```

```python
print(birthday.strftime("My birthday is %m/%d/%Y"))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)  # timedelta adds days across month and year boundaries
```

```python
print(ninety_d.strftime("90 days from today is %B %d, %Y"))
```

# Time for Dinner

```python
dinner_time = datetime.time(18, 30)
```

```python
print(dinner_time.strftime("Let's meet for dinner at %I:%M %p"))  # Use a 12-hour clock and show AM or PM
```
