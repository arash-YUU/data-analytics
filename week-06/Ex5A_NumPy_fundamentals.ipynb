Create arrays, inspect their data types, transpose an array, and select values.

```python
import numpy as np
```

## Array Basics

```python
fruits = ['apple', 'banana', 'cherry', 'date']
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)
# <U6 means Unicode text with room for up to 6 characters per item.
# U means Unicode, 6 is the character limit, and < shows the byte order.
```

```python
arr_2 = np.array([(10, 11, 12), (13, 14, 15)])
print(arr_2)
```

```python
print(arr_2.T)  # Rows become columns in the transposed view
```

```python
print(arr_2)
# Printing arr_2.T does not change the original array stored in arr_2.
```

```python
arr_3 = np.array([(1, 3, 5), (2, 4, 6)])
print(arr_3)
```

```python
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])  # Pick one value from each row using its row and column index
print(arr_4)
```

```python
arr_5 = np.ones((2, 2, 2))  # Two values along each of the three dimensions
print(arr_5)
```
