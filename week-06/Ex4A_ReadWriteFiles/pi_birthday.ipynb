```python
pi = open("pi_million_digits_bonus.txt", "r", encoding="utf-8")
pi_lines = pi.readlines()
pi.close()
```

```python
def pi_bday_check(bday):
    for line in pi_lines:
        if bday in line:
            print("Your birthday is in the first million digits of pi")
            break  # Stop searching after the first match
```

```python
pi_bday_check('063007')
```

## Give a result when no match is found

```python
def pi_bday_check2(bday):
    birthday_found = 0
    for line in pi_lines:
        if bday in line:
            print("Your birthday is in the first million digits of pi")
            birthday_found += 1
            break
    if birthday_found != 1:  # Print this once after the search, not once for every line
        print("Sorry, your birthday was not found in the first million digits of pi")
```

```python
for birthday in ['070281', '100799', '081505', '110252', '112901', '012411']:
    print(f"Checking {birthday}:")
    pi_bday_check2(birthday)
```

## Join the lines before finding the position
Searching the joined text also finds a birthday that crosses a line break.

```python
pi_string = ''
for line in pi_lines:
    pi_string += line.strip()  # Remove spaces and line breaks before joining the digits
print(pi_string[0:300])
```

```python
birthday_position = pi_string.index('063007') - 1
# Index 2 is the first decimal digit after 3., so subtracting 1 changes it to decimal place 1.
print(f"Your birthday begins at decimal place {birthday_position}")
```

```python
def pi_bday_check3(bday):
    bday_position = ''
    birthday_found = 0
    if bday in pi_string:  # The joined string includes matches across the original lines
        birthday_found = 1
    if birthday_found != 1:
        print("Sorry, your birthday was not found in the first million digits of pi")
    else:
        bday_position = pi_string.index(bday) - 1
        print("Your birthday is in the first million digits of pi")
        print(f"Your birthday begins at decimal place {bday_position}")
```

```python
for birthday in ['063007', '070281', '100799', '081505', '110252', '112901', '012411']:
    print(f"Checking {birthday}:")
    pi_bday_check3(birthday)
```
